```python
SEMANTIC_RULES = {
    # Waste-to-energy
    r'\b(wte|w2e|waste[-\s]?to[-\s]?energy)\b': 'waste-to-energy',

    # Anaerobic digestion
    r'\b(ad|biomethanation|anaerobic digestion)\b': 'anaerobic digestion',

    # Energy systems
    r'\b(chp|cogeneration|combined heat and power)\b': 'combined heat and power',

    # Biomass supply chain
    r'\b(bio[-\s]?supply chain|biomass supply chain management|bscm)\b': 'biomass supply chain',

    # Agricultural biomass
    r'\b(crop residues|agro[-\s]?residues)\b': 'agricultural biomass',
    r'\b(agro[-\s]?industrial waste)\b': 'agricultural residues',

    # Biomass types
    r'\b(cellulosic biomass|biomasses)\b': 'biomass',
    r'\b(woody biomass)\b': 'forest biomass',

    # GIS
    r'\b(gis|geospatial analysis|geographic information systems)\b': 'geographic information systems',

    # Optimization
    r'\b(optimization model|mathematical programming|optimal design)\b': 'mathematical optimization',
    r'\b(stochastic programming)\b': 'uncertainty analysis',

    # Decision-making
    r'\b(mcdm|madm|multi[-\s]?criteria decision making)\b': 'multiobjective optimization',

    # Multiobjective normalization
    r'\b(multi[-\s]?objective[s]? optimization|multiobjective optimization)\b': 'multiobjective optimization',

    # Development context
    r'\b(developing countries|global south|third world countries)\b': 'emerging economies',

    # Lemmatization-like fixes
    r'\b(optimisations|optimizing|optimisation)\b': 'optimization',
    r'\b(supply chains)\b': 'supply chain',
    r'\b(biofuels)\b': 'biofuel',
    r'\b(circular economies)\b': 'circular economy',

    # AI
    r'\b(ai)\b': 'artificial intelligence',
    r'\b(machine[-\s]?learning)\b': 'machine learning'
}
```

```python
def normalize_term(term, rules):
    term = term.lower().strip()
    for pattern, replacement in rules.items():
        if re.search(pattern, term, flags=re.IGNORECASE):
            return replacement
    return term
```

```python
bibfile.table_id_kwa['KWA_norm'] = bibfile.table_id_kwa['KWA'].apply(
    lambda x: normalize_term(x, SEMANTIC_RULES)
)
```

```python
bibfile.table_id_kwa = (
    bibfile.table_id_kwa
    .drop(columns=['KWA'])
    .rename(columns={'KWA_norm': 'KWA'})
    .drop_duplicates()
    .reset_index(drop=True)
)
```

```python
# Check Reference IDs (If References are Given)
#df = pd.DataFrame({'Reference': bibfile.u_ref, 'Reference ID': bibfile.u_ref_id})
#data_table.DataTable(df, num_rows_per_page = 10)

import bibtexparser
import re
import matplotlib.pyplot as plt


# === 1. Cargar archivo BibTeX ===
with open("/content/scopus3764.bib", encoding="utf-8") as bibtex_file:
    bib_database = bibtexparser.load(bibtex_file)

# === 2. Extraer datos relevantes ===
titulos = []
citas = []

for entry in bib_database.entries:
    titulos.append(entry.get("title", "Sin título"))
    note_field = entry.get("note", "")
    match = re.search(r"Cited by:\s*(\d+)", note_field)
    citas.append(int(match.group(1)) if match else 0)

# === 3. Crear DataFrame ===
df_citas = pd.DataFrame({
    "Título": titulos,
    "Citas": citas
})

# === 4. Ordenar por número de citas ===
df_citas = df_citas.sort_values(by="Citas", ascending=False)

# === 5. Visualizar top 10 artículos ===
plt.figure(figsize=(10, 6))
plt.barh(df_citas["Título"].head(10), df_citas["Citas"].head(10))
plt.xlabel("Número de Citas")
plt.ylabel("Artículo")
plt.title("Top 10 artículos más citados")
plt.gca().invert_yaxis()
plt.show()
```
