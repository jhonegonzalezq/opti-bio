```python
# Reemplazar valores vacíos o nulos "escondidos" por NaN reales
bibfile.data.replace(['', ' ', 'None', 'none', 'NaN', 'nan', 'null', 'NULL'], np.nan, inplace=True)

# Eliminar registros con al menos un campo faltante real
bibfile.data.dropna(how='any', inplace=True)

# Verificar nuevamente valores nulos
print("\n Nulos después de limpieza real:")
print(bibfile.data.isna().sum())

# Ver nuevo estado del health
print("\n Health después de limpieza REAL:")
print(bibfile.health_bib())
```

```python
# Define las columnas clave que quieres limpiar completamente
columnas_clave = ['abstract', 'doi', 'author_keywords']

# Reemplaza valores no válidos por np.nan
for col in columnas_clave:
    bibfile.data[col] = bibfile.data[col].apply(
        lambda x: np.nan if (
            pd.isna(x) or
            (isinstance(x, str) and str(x).strip().lower() in ['', 'none', 'nan', 'null']) or
            (isinstance(x, list) and len(x) == 0)
        ) else x
    )

# Elimina filas con np.nan en cualquiera de las columnas clave
bibfile.data.dropna(subset=columnas_clave, how='any', inplace=True)

# Verifica el estado del health después de limpieza
print("\n Health después de limpieza avanzada:")
print(bibfile.health_bib())
```

```python
# These indices are local (considers only the .bib scope)
# H-index measures their academic impact by identifying the number of papers (h) that have each received at least h citations
# E-Index quatifies excess citations within the H-core revealing "hidden" impact beyond the H-index threshold.
# The G-Index emphasizes highly cited work, making it sensitive to breakthrough publications.
# The M-Index contextualizes the H-index by normalizing it over the researcher’s career duration
aut_m = bibfile.m_index(2022)
df_idx = {
    'Author': bibfile.u_aut,
    'H-index': bibfile.aut_h,
    'E-Index': bibfile.aut_e,
    'G-Index': bibfile.aut_g,
    'M-Index': aut_m
}

df_idx = pd.DataFrame(df_idx)
df_idx
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# --- 1) Detectar columnas H/E de forma robusta ---
def norm(s):
    return s.strip().lower().replace(' ', '').replace('_','').replace('-','')

cols_map = {norm(c): c for c in df_idx.columns}

def pick(target):
    for k, orig in cols_map.items():
        if target in k:   # busca 'hindex' o 'eindex'
            return orig
    return None

h_col = pick('hindex')
e_col = pick('eindex')

if h_col is None or e_col is None:
    raise ValueError(f"No se encontraron columnas H/E. Columnas disponibles: {list(df_idx.columns)}")

# --- 2) Subconjunto y coerción a numérico ---
data = df_idx[[h_col, e_col]].copy()
data[h_col] = pd.to_numeric(data[h_col], errors='coerce')
data[e_col] = pd.to_numeric(data[e_col], errors='coerce')
data = data.dropna()

# --- 3) Scatter con fondo blanco ---
fig, ax = plt.subplots(figsize=(7,5), facecolor='white')
ax.scatter(data[h_col], data[e_col], s=32, alpha=0.8)
ax.set_xlabel('H-index')
ax.set_ylabel('E-Index')
ax.set_facecolor('white')
ax.grid(True, linewidth=0.6, alpha=0.4)
ax.spines[['top','right']].set_visible(False)
plt.show()
```

```python
# Check Reference IDs (If References are Given)

# df = pd.DataFrame({'Reference': bibfile.u_ref, 'Reference ID': bibfile.u_ref_id})
# data_table.DataTable(df, num_rows_per_page = 15)

import bibtexparser
import re
import pandas as pd
import matplotlib.pyplot as plt

# === 1. Cargar archivo BibTeX ===
with open("/content/scopus1792.bib", encoding="utf-8") as bibtex_file:
    bib_database = bibtexparser.load(bibtex_file)

# === 2. Extraer datos relevantes ===
titulos = []
citas = []

for entry in bib_database.entries:
    titulos.append(entry.get("title", "Sin título"))
    note_field = entry.get("note", "")
    match = re.search(r"Cited by:\s*(\d+)", note_field)
    citas.append(int(match.group(1)) if match else 0)

# === 3. Crear DataFrame ===
df_citas = pd.DataFrame({
    "Título": titulos,
    "Citas": citas
})

# === 4. Ordenar por número de citas ===
df_citas = df_citas.sort_values(by="Citas", ascending=False)

# === 5. Visualizar top 15 artículos ===
plt.figure(figsize=(8, 10))
plt.barh(df_citas["Título"].head(15), df_citas["Citas"].head(15))
plt.xlabel("Número de Citas")
plt.ylabel("Artículo")
plt.title("Top 15 artículos más citados")
plt.gca().invert_yaxis()
plt.subplots_adjust(left=0.3)
plt.show()
```

```python
# Check Table
num_columns = 5
data_wd     = bibfile.ask_gpt_wd
items       = list(data_wd.items())

field_names = []
for i in range(num_columns):
    field_names.append(f"Word {i + 1}")
    field_names.append(f"Importance {i + 1}")

table = PrettyTable()
table.field_names = field_names

for i in range(0, len(items), num_columns):
    row = []
    for j in range(num_columns):
        if i + j < len(items):
            word, importance = items[i + j]
            row.extend([word, round(importance, 4)])
        else:
            row.extend(["", ""])
    table.add_row(row)

# Print the table
print(table)
```
